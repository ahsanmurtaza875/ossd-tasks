```python
# ---------- Task 1 ----------
times = [120, 45, 32, 28, 41, 39, 35, 210]
result1 = times[-3:1:-1]          # start at 3rd from end, stop before index 1, step -1
print("Task 1:", result1)
assert result1 == [39, 41, 28, 32]
```

```python
# ---------- Task 2 ----------
tasks = ["done", "pending", "done", "pending", "done", "pending"]
tasks[1::2] = ["completed"] * len(tasks[1::2])   # replace every 2nd item from index 1
tasks[3:3] = ["audit"]                           # insert after the 3rd element
print("Task 2:", tasks)
assert tasks == ["done", "completed", "done", "audit", "completed", "done", "completed"]
```

```python
# ---------- Task 3 ----------
points = [(2, 5), (8, 3), (4, 4), (9, 1)]
result3 = [(y, x) if x > y else (x, y) for x, y in points]
print("Task 3:", result3)
assert result3 == [(2, 5), (3, 8), (4, 4), (1, 9)]
```

```python
# ---------- Task 4 ----------
raw_temps = [22.5, "ERR", 18.0, None, 30.2, "ERR", 0.0]
result4 = [round(c * 1.8 + 32, 1) for c in raw_temps if isinstance(c, (int, float))]
print("Task 4:", result4)
assert result4 == [72.5, 64.4, 86.4, 32.0]
```

```python
# ---------- Task 5 ----------
matrix = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
diagonal = [matrix[i][i] for i in range(len(matrix))]
evens = [num for row in matrix for num in row if num % 2 == 0]
print("Task 5 Diagonal:", diagonal)
print("Task 5 Evens:", evens)
assert diagonal == [1, 5, 9]
assert evens == [2, 4, 6, 8]
```

```python
# ---------- Task 6 ----------
prices = [100, 102, 104, 103, 105, 110]
k = 3
result6 = [(i, round(sum(prices[i:i + k]) / k, 2)) for i in range(len(prices) - k + 1)]
print("Task 6:", result6)
assert result6 == [(0, 102.0), (1, 103.0), (2, 104.0), (3, 106.0)]
```

```python
# ---------- Task 7 ----------
players = [("Aria", 88, True), ("Kael", 95, False), ("Zane", 91, True),
           ("Mira", 76, True), ("Lyra", 99, True), ("Dax", 84, True)]
active = [p for p in players if p[2]]                      # keep active only
top3 = sorted(active, key=lambda p: p[1], reverse=True)[:3]  # sort high->low, top 3
result7 = [f"Rank {n}: {name} ({score})" for n, (name, score, _) in enumerate(top3, start=1)]
print("Task 7:", result7)
assert result7 == ["Rank 1: Lyra (99)", "Rank 2: Zane (91)", "Rank 3: Aria (88)"]
```

```python
# ---------- Task 8 ----------
odometer = (1050, 1085, 1120, 1200, 1215)
result8 = [(start, end, end - start) for start, end in zip(odometer[:-1], odometer[1:])]
print("Task 8:", result8)
assert result8 == [(1050, 1085, 35), (1085, 1120, 35), (1120, 1200, 80), (1200, 1215, 15)]

print("\nAll tasks passed ✅")
```
